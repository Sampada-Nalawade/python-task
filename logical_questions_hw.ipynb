#### 1. Write a program to check whether a given number is even or odd.

```python
num = 27

if num % 2 == 0:
    print(num,"is even")
else:
    print(num,"is odd")
```

#### 2. Write a program to find the largest of two given numbers.

```python
num1 = 33
num2 = 77

if num1 == num2:
    print("Both are equal")
elif num1 > num2:
    print(num1,"is largest")
else:
    print(num2,"is largest")
```

#### 3. Write a program to count the number of digits in a given string.

```python
a = "Python123"

count = 0
for ch in a:
    if ch in '0123456789':
        count += 1

print("Total digits present:",count)
```

#### 4. Write a program to count how many uppercase and lowercase letters are present in a given string.

```python
s = 'Python 123 Programming'

lower_count = 0
upper_count = 0

for ch in s:
    if ch in 'ABCEDFGHIJKLMNOPQRSTUVWXYZ':
        upper_count += 1
    elif ch in 'abcdefghijklmnopqrstuvwxyz':
        lower_count += 1

print("Uppercase:",upper_count)
print("Lowercase:",lower_count)
```

#### 5. Write a program to reverse a given string without using any built-in reverse function.

```python
s = 'Python'

for ch in s:
    print(s[-1],end='')
    s = s[ :len(s)-1 : ]
```

#### 6. Write a program to find the sum of all numbers from 1 to a given number.

```python
n = 5

sum = 0
for i in range (1,n+1):
    sum += i
    i += 1

print("sum:",sum)
```

#### 7. Write a program to check whether a given string is a palindrome or not.

```python
s = 'level'
s_original = s
s_reverse = ''
for ch in s:
    s_reverse += s[-1]
    s = s[ :len(s)-1: ]

if s_original == s_reverse:
    print('Palindrome')
else:
    print('Not palindrome')
```
