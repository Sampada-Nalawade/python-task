### Operators in Python

### 1. Arithmetic Operators

```python
# int 
a = 30
b = 7
print(a+b) #addition
print(a-b) #subtraction
print(a*b) #multiplication
print(a/b) #division
print(a//b) #floor division
print(a%b) #modulo
```

```python
# float
c = 21.45
d = 29.77
print(c+d)
print(c-d) 
print(c*d)
print(c/d) 
print(c//d)
print(c%d)
```

```python
# complex
e = 2 + 4j
h = 9 - 7j
print(e+h)
print(e-h) 
print(e*h)
print(e/h)

# complex data type does not support floor division(//) and modulo(%) operations
```

```python
# bool
t = True
f = True
print(t+f)
print(t-f) 
print(t*f)
print(t/f) 
print(t//f)
print(t%f)
```

```python
# str
x = 'Python'
y = 'Programming'
print(x+y) #Concatenation

z = 5
print(x*z) #Repetition
```

### 2. Comparison (Relational) Operators

```python
# int
a = 55
b = 20
print(a>b)
print(a<b)
print(a>=b)
print(a<=b)
print(a==b)
print(a!=b)
```

```python
# float
a = 3.33
b = 7.2
print(a>b)
print(a<b)
print(a>=b)
print(a<=b)
print(a==b)
print(a!=b)
```

```python
# complex
a = 1+2j
b = 1+2j
print(a==b)
print(a!=b)

# complex data type does not support >, <, >=, <= comparison operators
```

```python
# bool
a = True
b = False
print(a>b)
print(a<b)
print(a>=b)
print(a<=b)
print(a==b)
print(a!=b)
```

```python
# str
a = 'Hello'
b = 'hey'
print(a>b)
print(a<b)
print(a>=b)
print(a<=b)
print(a==b)
print(a!=b)
```
