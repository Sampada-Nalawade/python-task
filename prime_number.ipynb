#### Homework : WAP to check given number is prime or not

```python
n = int(input('Enter a number:'))

is_prime = True

if n <= 1:
    print(n,'is not a prime number')
else:
    for i in range(2,n):
        if n % i == 0:
            is_prime = False

    if is_prime == True:
        print(n,'is a prime number')
    else:
        print(n,'is not a prime number')
```
