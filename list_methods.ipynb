#### Homework: WAP to print names of students who are absent today.

```python
Total_student = ['Raj','Amit','Jay','Payal','Priya','Rahul','Pavan','Kiran','Prachi']
present_attendance = ('Raj','Jay','Payal','Priya','Pavan','Kiran')

absent_student = []
for name in Total_student:
    if name not in present_attendance:
        absent_student.append(name)
# print(absent_student)
absent_stu = tuple(absent_student)
print(absent_stu)
```

#### List Methods:
#### 1. append

```python
l = [1,2,3]
print(l)
l.append(4)
print(l)
l.append([5,6]) # append takes exactly one element
print(l)
```

#### 2. extend

```python
l = [10,20,30]
print(l)
l.extend([40,50])
print(l)
```

#### 3. insert

```python
l = ['A','B','C','D']
print(l)
l.insert(2,'E')
print(l)
```

#### 4. remove

```python
l = [1,2,3,4,2]
print(l)
l.remove(2) # removes the first occurence
print(l)
```

#### 5. pop

```python
l = [10,20,30,40]
print(l)
p = l.pop() # remove and return last element by default 
print(l)
print(p)
pop = l.pop(1) # remove and return element at index 
print(l)
print(pop)
```

#### 6. clear

```python
l = [1,2,3,4]
print(l)
l.clear()
print(l)
```

#### 7. index

```python
l = ['a','b','c','d']
print(l)
print(l.index('c'))
```

#### 8. count

```python
l = [1,2,3,2,2,5,6,2]
print(l)
print(l.count(2))
```

#### 9. sort

```python
l = [7,3,2,5,4,1,9]
print(l)
l.sort() # ascending
print(l)
l.sort(reverse = True) #descending
print(l)
```

#### 10. reverse

```python
l = [2,4,1,6,3]
print(l)
l.reverse()
print(l)
```

#### 11. copy

```python
l = ['a','b','c']
print(l)
c = l.copy()
print(c)
```
